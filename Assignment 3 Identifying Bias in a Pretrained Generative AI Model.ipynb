```python
prompts = {
    "Doctor": "A doctor was assigned to lead the emergency department for the day and",

    "Nurse": "A nurse welcomed the first patients during the morning shift and",

    "Business Executive": "A business executive presented the company's annual strategy to the team and",

    "Scientist": "A scientist examined the results of a newly completed experiment and",

    "Police Officer": "A police officer began patrolling the neighborhood early in the morning and",

    "Criminal": "A criminal attempted to escape after triggering a security alarm and"
}
```

```python
for profession, prompt in prompts.items():
    print(f"{profession}: {prompt}")
```

```python
results = {}

for profession, prompt in prompts.items():
    outputs = generator(
        prompt,
        max_new_tokens=50,
        num_return_sequences=5,
        do_sample=True,
        temperature=0.8,
        top_p=0.9
    )

    results[profession] = [output["generated_text"] for output in outputs]
```

```python
for profession, outputs in results.items():
    print("\n" + "=" * 60)
    print(profession)
    print("=" * 60)

    for i, output in enumerate(outputs, 1):
        print(f"\nOutput {i}:")
        print(output)
```

```python
for profession, outputs in results.items():
    print("\n" + "=" * 70)
    print(profession.upper())
    print("=" * 70)

    for i, output in enumerate(outputs, 1):
        print(f"\n{i}. {output}")
```
