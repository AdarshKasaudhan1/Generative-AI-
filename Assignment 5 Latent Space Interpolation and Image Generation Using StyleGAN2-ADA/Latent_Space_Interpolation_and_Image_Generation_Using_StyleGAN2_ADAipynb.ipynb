```python
import torch
import pickle
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import os
import imageio
```

```python
import urllib.request

url = "https://nvlabs-fi-cdn.nvidia.com/stylegan2-ada-pytorch/pretrained/ffhq.pkl"

urllib.request.urlretrieve(url, "ffhq.pkl")

print("Model downloaded successfully!")
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

with open("ffhq.pkl", "rb") as f:
    G = pickle.load(f)["G_ema"].to(device)

print("Device:", device)
print("Latent dimension:", G.z_dim)
```

```python
torch.manual_seed(42)

z1 = torch.randn([1, G.z_dim], device=device)
z2 = torch.randn([1, G.z_dim], device=device)

print("Shape of z1:", z1.shape)
print("Shape of z2:", z2.shape)
```

```python
def generate_image(z):
    with torch.no_grad():
        img = G(z, None, truncation_psi=0.7, noise_mode="const")

    img = (img.clamp(-1, 1) + 1) / 2
    img = img[0].permute(1, 2, 0).cpu().numpy()

    return img


img1 = generate_image(z1)
img2 = generate_image(z2)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img1)
plt.title("Image 1 - z₁")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img2)
plt.title("Image 2 - z₂")
plt.axis("off")

plt.show()
```

```python
num_images = 12

alphas = np.linspace(0, 1, num_images)

interpolated_images = []

for alpha in alphas:
    z = (1 - alpha) * z1 + alpha * z2
    img = generate_image(z)
    interpolated_images.append(img)
```

```python
fig, axes = plt.subplots(1, num_images, figsize=(24, 4))

for i, (ax, img) in enumerate(zip(axes, interpolated_images)):
    ax.imshow(img)
    ax.set_title(f"{i+1}")
    ax.axis("off")

plt.tight_layout()
plt.savefig("stylegan_interpolation_strip.png", dpi=200, bbox_inches="tight")
plt.show()
```

```python
frames = []

for img in interpolated_images:
    frame = (img * 255).astype(np.uint8)
    frames.append(frame)

imageio.mimsave(
    "stylegan_interpolation.gif",
    frames,
    duration=0.25,
    loop=0
)

print("GIF created successfully!")
```

```python
z_base = z1.clone()

values = np.linspace(-3, 3, 10)

dimension_images = []

for value in values:

    z_modified = z_base.clone()

    # Change ONLY dimension 0
    z_modified[0, 0] = value

    img = generate_image(z_modified)

    dimension_images.append(img)
```

```python
fig, axes = plt.subplots(1, 10, figsize=(25, 4))

for i, (ax, img) in enumerate(zip(axes, dimension_images)):
    ax.imshow(img)
    ax.set_title(f"{values[i]:.1f}")
    ax.axis("off")

plt.tight_layout()
plt.savefig(
    "latent_dimension_experiment.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()
```

###Observation


The experiment demonstrated how StyleGAN2-ADA generates face images from latent vectors. By interpolating between two randomly generated latent vectors, a sequence of images was produced to illustrate the transition between the generated faces. Additionally, modifying the first latent dimension across different values produced variations in the generated images. This experiment helped demonstrate how latent vectors control image generation and how changes in latent-space values can influence the resulting facial appearance.
